# Task 3: A/B Hypothesis Testing

This notebook validates risk differences using claim frequency and margin. It uses the local repository dataset, whose schema differs from the original challenge brief.

```python
from src.data_loader import load_data
from src.hypothesis_tests import chi_square_claim_frequency, decision, two_proportion_z_test, welch_t_test

df = load_data('../data/insurance_data.csv')
df.head()
```

```python
tests = []

province = chi_square_claim_frequency(df, 'Province')
tests.append({'Hypothesis': 'No risk difference across provinces', 'KPI': 'Claim Frequency', 'Test': province['test'], 'p_value': province['p_value'], 'Decision': decision(province['p_value'])})

zip_risk = two_proportion_z_test(df, 'ZipCode', 30002, 40002)
tests.append({'Hypothesis': 'No risk difference between zip codes', 'KPI': 'Claim Frequency', 'Test': zip_risk['test'], 'p_value': zip_risk['p_value'], 'Decision': decision(zip_risk['p_value'])})

zip_margin = welch_t_test(df, 'ZipCode', 30002, 40002, metric='Margin')
tests.append({'Hypothesis': 'No margin difference between zip codes', 'KPI': 'Margin', 'Test': zip_margin['test'], 'p_value': zip_margin['p_value'], 'Decision': decision(zip_margin['p_value'])})

gender = two_proportion_z_test(df, 'Gender', 'Female', 'Male')
tests.append({'Hypothesis': 'No risk difference between Women and Men', 'KPI': 'Claim Frequency', 'Test': gender['test'], 'p_value': gender['p_value'], 'Decision': decision(gender['p_value'])})

import pandas as pd
pd.DataFrame(tests)
```
